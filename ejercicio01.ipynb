```python

# Ejercicio 1.1 - Funciones Lambda

# a. Recibir un número y determinar si es múltiplo de 3
es_multiplo_de_tres = lambda numero: numero % 3 == 0

# b. Recibir un número y devolver su cubo
calcular_cubo = lambda numero: numero ** 3

# c. Recibir dos números y devolver su producto
calcular_producto = lambda numero1, numero2: numero1 * numero2

# d. Recibir dos números y devolver el mayor
obtener_mayor = lambda numero1, numero2: max(numero1, numero2)

# e. Recibir una palabra y determinar si empieza por A o a
empieza_con_a = lambda palabra: palabra.lower().startswith("a")

# f. Convertir una temperatura de Celsius a Fahrenheit
convertir_a_fahrenheit = lambda celsius: (celsius * 9/5) + 32


# Pruebas de funcionamiento

print("a. ¿Es múltiplo de 3?", es_multiplo_de_tres(9))
print("b. Cubo:", calcular_cubo(3))
print("c. Producto:", calcular_producto(4, 5))
print("d. Mayor:", obtener_mayor(8, 12))
print("e. ¿Empieza con A?", empieza_con_a("Arbol"))
print("f. Temperatura en Fahrenheit:", convertir_a_fahrenheit(25))


# Ejercicio 1.2 - Lista de funciones Lambda

# Crear una lista con tres funciones lambda
funciones = [
    lambda numero: numero * 2,
    lambda numero: numero + 10,
    lambda numero: numero ** 2
]

# Solicitar un número al usuario
numero = float(input("Ingrese un número: "))

# Aplicar cada función al número ingresado
for funcion in funciones:
    print(funcion(numero))
```
