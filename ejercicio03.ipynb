```python

# a. Convertir temperaturas de Celsius a Fahrenheit

temperaturas_celsius = [0, 10, 20, 25, 30, 35]

temperaturas_fahrenheit = list(map(lambda c: (c * 9/5) + 32, temperaturas_celsius))

print("a. Temperaturas en Fahrenheit:", temperaturas_fahrenheit)


# b. Sumar los elementos de dos listas

lista1 = [1, 2, 3, 4, 5]
lista2 = [10, 20, 30, 40, 50]

resultado_b = list(map(lambda a, b: a + b, lista1, lista2))

print("b. Suma de las dos listas:", resultado_b)


# c. Calcular las areas de una lista de radios de circulos

radios = [1, 2, 3, 4, 5]

areas = list(map(lambda radio: 3.1416 * radio**2, radios))

print("c. Areas de los circulos:", areas)


# d. Generar una lista de tuplas (nombre, valor + 10%)

productos = {
    "pan": 1000,
    "leche": 2500,
    "cafe": 5000
}

resultado_d = list(map(lambda producto: (producto[0], producto[1] * 1.10), productos.items()))

print("d. Productos con aumento del 10%:", resultado_d)


# e. Multiplicar todos los elementos de una matriz por 10

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

resultado_e = list(map(lambda fila: list(map(lambda numero: numero * 10, fila)), matriz))

print("e. Matriz multiplicada por 10:", resultado_e)
```
