```python

# Ejercicio 8 - Minimax: Juego NIM

# Obtener los movimientos posibles
def obtener_movimientos(piedras):
    return list(range(1, min(3, piedras) + 1))


# Algoritmo Minimax
def minimax(piedras, turno):
    # Si no quedan piedras, pierde el jugador que tiene el turno
    if piedras == 0:
        if turno == "MAX":
            return -1
        else:
            return 1

    # Obtener los movimientos posibles
    movimientos = obtener_movimientos(piedras)

    # Turno de la computadora (MAX)
    if turno == "MAX":
        mejor_valor = -float("inf")

        for movimiento in movimientos:
            valor = minimax(piedras - movimiento, "MIN")
            mejor_valor = max(mejor_valor, valor)

        return mejor_valor

    # Turno del humano (MIN)
    else:
        mejor_valor = float("inf")

        for movimiento in movimientos:
            valor = minimax(piedras - movimiento, "MAX")
            mejor_valor = min(mejor_valor, valor)

        return mejor_valor


# Elegir el mejor movimiento para la computadora
def mejor_movimiento(piedras):
    movimientos = obtener_movimientos(piedras)

    mejor_valor = -float("inf")
    mejor_jugada = movimientos[0]

    for movimiento in movimientos:
        valor = minimax(piedras - movimiento, "MIN")

        if valor > mejor_valor:
            mejor_valor = valor
            mejor_jugada = movimiento

    return mejor_jugada


# Simulacion del juego NIM
def jugar_nim():
    piedras = 10
    turno = "MIN"

    print("================================")
    print("       JUEGO NIM - MINIMAX")
    print("================================")
    print("Hay 10 piedras en la pila.")
    print("Puedes retirar entre 1 y 3 piedras.")
    print("Tu eres MIN y juegas primero.")
    print("La computadora es MAX.")
    print("================================")

    while piedras > 0:

        print("\nPiedras restantes:", piedras)
        print("O" * piedras)

        if turno == "MIN":
            print("\nTurno del jugador (MIN)")

            while True:
                try:
                    movimiento = int(input("¿Cuantas piedras deseas retirar? (1-3): "))

                    if movimiento in obtener_movimientos(piedras):
                        break
                    else:
                        print("Movimiento no valido. Intenta nuevamente.")

                except ValueError:
                    print("Debes ingresar un numero entero.")

        else:
            print("\nTurno de la computadora (MAX)")

            movimiento = mejor_movimiento(piedras)

            print("La computadora retira", movimiento, "piedra(s).")

        piedras -= movimiento

        print("Piedras retiradas:", movimiento)
        print("Piedras restantes:", piedras)

        # Cambiar el turno
        if turno == "MIN":
            turno = "MAX"
        else:
            turno = "MIN"

    # El jugador que recibe el turno sin piedras pierde
    if turno == "MIN":
        print("\n¡La computadora gana! (MAX)")
        print("Te quedaste sin movimientos.")
    else:
        print("\n¡Felicidades! Has ganado. (MIN)")
        print("La computadora se quedo sin movimientos.")


# Iniciar el juego
jugar_nim()
```
