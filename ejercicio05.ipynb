```python

from functools import reduce


# a. Calcular la multiplicacion de todos los numeros

numeros = [2, 3, 4, 5]

resultado_a = reduce(lambda a, b: a * b, numeros)

print("a. Multiplicacion de los numeros:", resultado_a)


# b. Concatenar una lista de palabras en una sola cadena

palabras = ["Hola", "mundo", "desde", "Python"]

resultado_b = reduce(lambda a, b: a + " " + b, palabras)

print("b. Palabras concatenadas:", resultado_b)


# c. Encontrar el mayor numero utilizando una funcion auxiliar

def obtener_mayor(a, b):
    if a > b:
        return a
    else:
        return b


numeros = [15, 42, 8, 67, 23, 91, 34]

resultado_c = reduce(obtener_mayor, numeros)

print("c. El mayor numero es:", resultado_c)
```
