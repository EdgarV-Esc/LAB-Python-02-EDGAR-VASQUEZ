```python

# a. Ordenar una lista de palabras alfabeticamente

palabras = ["pera", "manzana", "uva", "banano", "naranja"]

resultado_a = sorted(palabras)

print("a. Palabras ordenadas alfabeticamente:", resultado_a)


# b. Ordenar una lista de tuplas por el primer elemento

tuplas = [(3, "Carlos"), (1, "Ana"), (4, "Pedro"), (2, "Maria")]

resultado_b = sorted(tuplas, key=lambda elemento: elemento[0])

print("b. Tuplas ordenadas por el primer elemento:", resultado_b)


# c. Ordenar productos por precio de mayor a menor y luego por nombre

productos = [
    ("Camisa", 30),
    ("Pantalon", 55),
    ("Medias", 10),
    ("Chaqueta", 80),
    ("Zapatos", 55)
]

resultado_c = sorted(productos, key=lambda producto: (-producto[1], producto[0]))

print("c. Productos ordenados por precio y nombre:", resultado_c)


# d. Ordenar palabras segun su longitud

palabras = ["programacion", "sol", "computador", "casa", "Python"]

resultado_d = sorted(palabras, key=lambda palabra: len(palabra))

print("d. Palabras ordenadas por longitud:", resultado_d)


# e. Ordenar numeros colocando primero los pares y luego los impares

numeros = [7, 2, 9, 4, 1, 8, 5, 6, 3]

resultado_e = sorted(numeros, key=lambda numero: (numero % 2, numero))

print("e. Numeros pares primero e impares despues:", resultado_e)


# f. Contar la aparicion de cada letra y ordenar por frecuencia

palabra = "programacion"

frecuencias = {
    letra: palabra.count(letra)
    for letra in set(palabra)
}

resultado_f = dict(sorted(frecuencias.items(), key=lambda elemento: elemento[1]))

print("f. Frecuencia de cada letra ordenada:", resultado_f)


# g. Ordenar empleados primero por departamento y luego por edad

empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnologia", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnologia", 28)
]

resultado_g = sorted(empleados, key=lambda empleado: (empleado[1], empleado[2]))

print("g. Empleados ordenados por departamento y edad:", resultado_g)
```
