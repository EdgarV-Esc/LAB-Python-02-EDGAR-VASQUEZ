```python

# a. Filtrar palabras con mas de 4 letras

palabras = ["sol", "perro", "gato", "computador", "casa", "programacion"]

resultado_a = list(filter(lambda palabra: len(palabra) > 4, palabras))

print("a. Palabras con mas de 4 letras:", resultado_a)


# b. Filtrar los elementos que no son nulos (None)

elementos = [10, None, 20, None, 30, 40, None, 50]

resultado_b = list(filter(lambda elemento: elemento is not None, elementos))

print("b. Elementos diferentes de None:", resultado_b)


# c. Filtrar palabras que empiezan por una vocal

palabras = ["Arbol", "casa", "elefante", "oso", "perro", "uva", "isla"]

resultado_c = list(filter(lambda palabra: palabra[0].lower() in "aeiou", palabras))

print("c. Palabras que empiezan por vocal:", resultado_c)


# d. Filtrar palabras que son palindromos

palabras = ["radar", "casa", "oso", "reconocer", "perro", "ana"]

resultado_d = list(filter(lambda palabra: palabra.lower() == palabra.lower()[::-1], palabras))

print("d. Palabras palindromas:", resultado_d)


# e. Filtrar numeros que terminan en 5

numeros = [15, 20, 35, 42, 55, 60, 75, 81, 95]

resultado_e = list(filter(lambda numero: numero % 10 == 5, numeros))

print("e. Numeros que terminan en 5:", resultado_e)


# f. Filtrar productos con precio mayor a 100

productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

resultado_f = list(filter(lambda producto: producto["precio"] > 100, productos))

print("f. Productos con precio mayor a 100:", resultado_f)
```
